```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [10000, 15000, 12000, 20000, 25000, 30000],
    'Profit': [2000, 3000, 2500, 5000, 6000, 8000],
    'Customers': [100, 150, 120, 200, 250, 300]
}

df = pd.DataFrame(data)

print(df)
plt.figure(figsize=(8, 5))

plt.plot(df['Month'], df['Sales'],
         marker='o', color='blue')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.grid(True)
plt.show()
plt.figure(figsize=(8, 5))

plt.bar(df['Month'], df['Sales'], color='green')

plt.title('Monthly Sales Comparison')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
plt.figure(figsize=(8, 5))

plt.scatter(df['Customers'], df['Sales'],
            color='red')

plt.title('Customers vs Sales')
plt.xlabel('Customers')
plt.ylabel('Sales')

plt.show()
plt.figure(figsize=(8, 5))

plt.hist(df['Sales'], bins=5, color='orange',
         edgecolor='black')

plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
plt.figure(figsize=(8, 5))

plt.boxplot(df['Sales'])

plt.title('Sales Distribution Box Plot')
plt.ylabel('Sales')

plt.show()
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='Month', y='Sales')

plt.title('Monthly Sales Analysis')
plt.show()
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=df,
    x='Customers',
    y='Sales'
)

plt.title('Customer and Sales Relationship')
plt.show()
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Line Chart
plt.figure(figsize=(8, 5))
plt.plot(df['Month'], df['Sales'], marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('line_chart.png', dpi=300, bbox_inches='tight')
plt.show()

# 2. Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(df['Month'], df['Sales'], color='green')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()

# 3. Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(df['Customers'], df['Sales'], color='red')
plt.title('Customers vs Sales')
plt.xlabel('Customers')
plt.ylabel('Sales')
plt.savefig('scatter_plot.png', dpi=300, bbox_inches='tight')
plt.show()

# 4. Histogram
plt.figure(figsize=(8, 5))
plt.hist(df['Sales'], bins=5, color='orange', edgecolor='black')
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.savefig('histogram.png', dpi=300, bbox_inches='tight')
plt.show()

# 5. Box Plot
plt.figure(figsize=(8, 5))
plt.boxplot(df['Sales'])
plt.title('Sales Box Plot')
plt.ylabel('Sales')
plt.savefig('box_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
