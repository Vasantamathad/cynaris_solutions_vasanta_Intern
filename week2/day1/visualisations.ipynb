```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [10000, 15000, 12000, 20000, 25000, 30000],
    'Profit': [2000, 3000, 2500, 5000, 6000, 8000],
    'Customers': [100, 150, 120, 200, 250, 300]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df['Month'], df['Sales'],
         marker='o', color='blue')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.grid(True)
plt.show()
```
