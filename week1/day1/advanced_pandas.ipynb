```python

import pandas as pd

# Customer dataset
customers = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 4],
    'Name': ['Rahul', 'Priya', 'Amit', 'Sneha']
})

# Orders dataset
orders = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 5],
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Keyboard'],
    'Amount': [50000, 20000, 15000, 3000]
})

print("CUSTOMERS DATA")
print(customers)

print("\nORDERS DATA")
print(orders)
```

```python
inner = pd.merge(customers, orders, on='Customer_ID', how='inner')
print(inner)
```

```python
left = pd.merge(customers, orders, on='Customer_ID', how='left')
print(left)
```

```python
right = pd.merge(customers, orders, on='Customer_ID', how='right')
print(right)
```

```python
outer = pd.merge(customers, orders, on='Customer_ID', how='outer')
print(outer)
```

```python

sales = pd.DataFrame({
    'Region': ['North', 'South', 'North', 'South',
               'East', 'East'],
    'Product': ['Laptop', 'Laptop', 'Mobile',
                'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 60000, 20000, 25000, 55000, 30000]
})

print(sales)
```

```python

pivot = pd.pivot_table(
    sales,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print("REGION-WISE PRODUCT SALES")
print(pivot)
```

```python

wide_data = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Tablet'],
    'January': [50000, 20000, 15000],
    'February': [55000, 25000, 18000],
    'March': [60000, 30000, 20000]
})

print("WIDE FORMAT DATA")
print(wide_data)
```

```python

long_data = pd.melt(
    wide_data,
    id_vars=['Product'],
    var_name='Month',
    value_name='Sales'
)

print("LONG FORMAT DATA")
print(long_data)
```

```python

employees = pd.DataFrame({
    'Name': ['Rahul', 'Priya', 'Amit', 'Sneha', 'Kiran'],
    'Department': ['IT', 'HR', 'IT', 'HR', 'Finance'],
    'Salary': [50000, 40000, 60000, 45000, 35000]
})

print(employees)
```

```python

result = (
    employees
    .query('Salary > 40000')
    .sort_values(by='Salary', ascending=False)
    .reset_index(drop=True)
)

print("FILTERED AND SORTED EMPLOYEE DATA")
print(result)
```
